```python
import os

# 확인할 경로
path = "./Dataset"

# 카운트 변수
jpg_count = 0
txt_count = 0

# 디렉토리 순회
for root, dirs, files in os.walk(path):
    for file in files:
        if file.lower().endswith(".jpg"):
            jpg_count += 1
        elif file.lower().endswith(".txt"):
            txt_count += 1

print(f".jpg 파일 개수: {jpg_count}")
print(f".txt 파일 개수: {txt_count}")
```
